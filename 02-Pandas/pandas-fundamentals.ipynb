```python
df["FamilyCategory"] = pd.cut(
    df["FamilySize"],
    bins=[0, 1, 4, float("inf")],
    labels=["Alone", "Small Family", "Large Family"]
)
```

```python
test = pd.Series(["Noob", "Beginner", "Pro", "World Class", "YOLO"])

test = test.map({
    "Noob" : 1,
    "Beginner" : 2,
    "Pro" : 3,
    "World Class" : 4
})

print(test)
```

```python
def categorize_Age(age):
    if age < 18:
        return "Child"
    elif age < 60:
        return "Adult"
    else:
        return "Senior"


df["AgeCategory"] = df["Age"].apply(categorize_Age)
```
