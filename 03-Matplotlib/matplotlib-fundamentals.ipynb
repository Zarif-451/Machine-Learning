# Bar Chart

A **Bar Chart** is a graphical representation used to compare **categorical data**. It displays categories along one axis and their corresponding values or frequencies as rectangular bars, where the height (or length) of each bar is proportional to the represented value.

Bar charts are widely used in **Exploratory Data Analysis (EDA)** to visualize and compare the frequency or count of different categories, making it easier to identify patterns, differences, and distributions within a dataset.

**Examples:** Number of male vs. female passengers, passenger class distribution, and survival counts in the Titanic dataset.

| **Line Plot (`plt.plot()`)** | **Bar Chart (`plt.bar()`)** |
|:----------------------------:|:---------------------------:|
| Continuous data | Categorical data |
| Shows trends over time | Compares categories |
| Connects data points with lines | Represents values using separate bars |

````python
gender_counts = df["Sex"].value_counts()
````

# Subplots

## Pyplot vs Object-Oriented Approach in Matplotlib

Matplotlib provides two common ways to create and customize visualizations:

1. **Pyplot approach** – Uses `plt` functions directly.
2. **Object-oriented approach** – Uses `Figure` and `Axes` objects (`fig` and `ax`).

Both approaches can produce the same visualizations, but the object-oriented approach provides better control, especially when working with multiple plots.

### 1. Understanding the difference

| Pyplot approach | Object-oriented approach |
|---|---|
| `plt.plot()` | `ax.plot()` |
| `plt.hist()` | `ax.hist()` |
| `plt.bar()` | `ax.bar()` |
| `plt.scatter()` | `ax.scatter()` |
| `plt.title()` | `ax.set_title()` |
| `plt.xlabel()` | `ax.set_xlabel()` |
| `plt.ylabel()` | `ax.set_ylabel()` |
| `plt.legend()` | `ax.legend()` |
| `plt.grid()` | `ax.grid()` |

Notice that in the object-oriented approach, most customization methods are called directly on the `ax` object. For example, `plt.title()` becomes `ax.set_title()`.

### 2. Understanding Figure and Axes

Before using the object-oriented approach, we need to understand two important objects:

- **Figure (`fig`):** The entire canvas or container that holds our visualizations.
- **Axes (`ax`):** The individual plotting area where we draw graphs.

Think of a Figure as a sheet of paper and an Axes as a graph drawn on that paper.

A Figure can contain one or multiple Axes.

### 3. Practical implementation

**Using the Pyplot approach:**

```python
plt.figure(figsize=(8, 5))

plt.hist(df["Age"], bins=20)
plt.title("Age Distribution")
plt.xlabel("Age")
plt.ylabel("Frequency")
plt.grid()

plt.show()
```

**Using the Object-Oriented approach:**

```python
fig, ax = plt.subplots(figsize=(8, 5))

ax.hist(df["Age"], bins=20)
ax.set_title("Age Distribution")
ax.set_xlabel("Age")
ax.set_ylabel("Frequency")
ax.grid()

plt.show()
```

Both examples produce essentially the same visualization.

### 4. Why prefer the Object-Oriented approach?

The object-oriented approach becomes particularly useful when creating multiple plots.

For example:

```python
fig, ax = plt.subplots(1, 2, figsize=(12, 5))

ax[0].hist(df["Age"], bins=20)
ax[0].set_title("Age Distribution")

ax[1].hist(df["Fare"], bins=20)
ax[1].set_title("Fare Distribution")

plt.tight_layout()
plt.show()
```

Here:
- `ax[0]` controls the first plotting area.
- `ax[1]` controls the second plotting area.
- Each Axes object can be customized independently.

This makes the code easier to organize and maintain as visualizations become more complex.

### Key Takeaways

- `plt` provides a convenient interface for creating plots.
- `fig` represents the entire figure.
- `ax` represents an individual plotting area.
- A Figure can contain multiple Axes.
- The object-oriented approach offers more flexibility and control, particularly when working with subplots.

**Recommendation:** Become comfortable with both approaches, but prefer the object-oriented approach (`fig`, `ax`) when creating ML visualizations and multiple subplots.

````python
fig, ax = plt.subplots(1, 3, figsize=(15, 5))

ax[0].hist(df["Age"])
ax[0].set_title("Age Distribution")

ax[1].hist(df["Fare"])
ax[1].set_title("Fare Distribution")

survival_counts = df["Survived"].value_counts()
ax[2].bar(survival_counts.index, survival_counts.values)


plt.tight_layout()
````

# Axis Customization in Matplotlib

## Axis Limits in Matplotlib

Axis limits control the visible range of a plot along the X-axis and Y-axis.

### Functions

| Function | Purpose |
|---|---|
| `plt.xlim()` | Set X-axis limits |
| `plt.ylim()` | Set Y-axis limits |
| `ax.set_xlim()` | Set X-axis limits (Object-Oriented) |
| `ax.set_ylim()` | Set Y-axis limits (Object-Oriented) |

### Example

```python
ax.set_xlim(0, 80)
ax.set_ylim(0, 100)
```

- `ax.set_xlim(0, 80)`: Displays the X-axis from 0 to 80.
- `ax.set_ylim(0, 100)`: Displays the Y-axis from 0 to 100.

### Important Notes

- Axis limits change the visible plotting range, not the actual dataset.
- Values outside the specified range are hidden from the visualization, not deleted.
- Useful for zooming into specific regions and making patterns easier to observe.

````python
plt.scatter(df["Age"], df["Fare"])

plt.xlim(0, 80)
plt.ylim(0, 100)
plt.grid()
````

````python
fig, ax = plt.subplots()

ax.scatter(df["Age"], df["Fare"])
ax.set_xlim(0, 80)
ax.set_ylim(0, 100)

ax.set_title("Age Vs. Fare")
ax.set_xlabel("Age")
ax.set_ylabel("Fare")
plt.grid()
````

## Axis Ticks 

````python
fig, ax = plt.subplots()
ax.scatter(df["Age"], df["Fare"], alpha=0.5)

ax.set_xticks(range(0, 91, 10))
ax.set_ylim(0, 100)
ax.set_xlabel("Age")
ax.set_ylabel("Fare")
ax.set_title("Age Vs. Fare")

plt.grid()
````

````python
fig, ax = plt.subplots()

ax.bar(["First", "Second", "Third"], [216, 184, 491])

ax.set_xticks([0, 1, 2])
ax.set_xticklabels(["1st Class", "2nd Class", "3rd Class"])

plt.show()
````

## Transparency (`alpha`) in Matplotlib

The `alpha` parameter controls the transparency of plotted elements.

### Alpha Values

| Alpha | Appearance |
|---|---|
| `0` | Completely transparent (invisible) |
| `0.3` | Mostly transparent |
| `0.5` | 50% transparent |
| `1` | Completely opaque (fully visible) |

### Example

```python
plt.scatter(df["Age"], df["Fare"], alpha=0.3)
```

Here, `alpha=0.3` makes the scatter plot points partially transparent.

### Why Is It Useful?

- Helps visualize overlapping points.
- Makes dense areas easier to identify.
- Useful when plotting datasets with many observations.

**Note:** Lower `alpha` values mean more transparency; higher values mean less transparency.

````python
plt.scatter(df["Age"], df["Fare"], alpha=0.3)
````

## Logarithmic Scale in Matplotlib

A logarithmic scale is useful when numerical values vary significantly in magnitude.

### 1. Linear vs Logarithmic Scale

- **Linear scale:** Equal distances represent equal numerical differences.
- **Logarithmic scale:** Equal distances represent equal multiplicative changes.

For example, on a logarithmic scale, `1`, `10`, `100`, and `1000` are equally spaced because each value is 10 times the previous one.

### 2. Matplotlib Functions

| Function | Purpose |
|---|---|
| `plt.xscale("log")` | Set X-axis to logarithmic scale |
| `plt.yscale("log")` | Set Y-axis to logarithmic scale |
| `ax.set_xscale("log")` | Set X-axis to logarithmic scale (Object-Oriented) |
| `ax.set_yscale("log")` | Set Y-axis to logarithmic scale (Object-Oriented) |

### 3. Example

```python
fig, ax = plt.subplots(figsize=(8, 5))

ax.scatter(df["Age"], df["Fare"], alpha=0.4)

ax.set_yscale("log")

ax.set_xlabel("Age")
ax.set_ylabel("Fare (Log Scale)")
ax.set_title("Age vs Fare")

plt.show()
```

Here, the Y-axis uses a logarithmic scale, making it easier to visualize values that vary significantly.

### 4. Important Notes

- Useful when values span several orders of magnitude.
- Changes how values are displayed, not the original dataset.
- Standard logarithmic scales require positive values. Zero and negative values need special handling.

````python
fig, ax = plt.subplots(figsize=(8, 5))

ax.scatter(df["Age"], df["Fare"], alpha=0.4)

ax.set_yscale("log")

ax.set_xlabel("Age")
ax.set_ylabel("Fare (Log Scale)")
ax.set_title("Age vs Fare")

plt.show()
````
