## Operations on Vectors 

```python
#vector indexing operations on 1-D vectors
a = np.arange(10)
print(a)

#access an element
print(f"a[2].shape: {a[2].shape} a[2]  = {a[2]}, Accessing an element returns a scalar")

# access the last element, negative indexes count from the end
print(f"a[-1] = {a[-1]}")

#indexs must be within the range of the vector or they will produce and error
try:
    c = a[10]
except Exception as e:
    print("The error message you'll see is:")
    print(e)
```

## Slicing 

### Slicing creates an array of indices using a set of three values (start:stop:step). A subset of values is also valid. Its use is best explained by example:  

```python
#vector slicing operations
a = np.arange(10)
print(f"a         = {a}")

#access 5 consecutive elements (start:stop:step)
c = a[2:7:1];     print("a[2:7:1] = ", c)

# access 3 elements separated by two 
c = a[2:7:2];     print("a[2:7:2] = ", c)

# access all elements index 3 and above
c = a[3:];        print("a[3:]    = ", c)

# access all elements below index 3
c = a[:3];        print("a[:3]    = ", c)

# access all elements
c = a[:];         print("a[:]     = ", c)
```

## Vector Vector Dot Product  

```python
np.random.seed(1)    #SAME RANDOM VALUE EVERY TIME

a = np.random.rand(10000000)
b = np.random.rand(10000000)

tic = time.time()        #start time
c = np.dot(a, b)
toc = time.time()        #end time

print(f"np.dot(a, b) : {c}")
print(f"Start time : {tic}")
print(f"End time : {toc}")
print(f"Total Execution Time : {1000*(toc - tic):.4f} ms")

del(a);del(b)           #removes the array from memory
```

## Operations on Matrices 

## Indexing 

```python
a = np.arange(6).reshape(-1, 2)      #-1 -> NumPy figures out the dimension 
                                    #for row
print(f"Shape of Matrix a : {a.shape}")
print(a)

#access an element
print(f"a[2, 0] = {a[2, 0]}")

#access a row
print(f"a[2] = {a[2]}")
```

## Slicing

```python
a = np.arange(20).reshape(-1, 10)
print("Original Matrix : ")
print(a)
print("\n\n")

#access 5 consecutive elements (start:stop:step)
b = a[0, 2:7:1]
print(b)
print(b.shape)

#access all elements
b = a[:,:]
print(b)
print(b.shape)

#access all elements in 1 row
b = a[0,:]
print(b)
print(b.shape)
```
