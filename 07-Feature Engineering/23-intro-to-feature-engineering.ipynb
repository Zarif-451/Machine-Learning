# Feature Engineering

Feature engineering is the process of using **domain knowledge** to extract features from raw data.

These features can be used to improve the performance of **Machine Learning algorithms**.

---

## Machine Learning Workflow

$$
\text{Datasets}
\rightarrow
\text{Data Retrieval}
\rightarrow
\text{Data Preparation}
\rightarrow
\text{Modeling}
\rightarrow
\text{Model Evaluation \& Tuning}
\rightarrow
\text{Deployment \& Monitoring}
$$

### Data Preparation

Data Preparation consists of:

1. **Data Processing & Wrangling**
2. **Feature Extraction & Engineering**
3. **Feature Scaling & Selection**

These steps prepare the data before it is given to the Machine Learning model.

---

## Complete Workflow

$$
\text{Datasets}
\rightarrow
\text{Data Retrieval}
\rightarrow
\begin{cases}
\text{Data Processing \& Wrangling}\\
\text{Feature Extraction \& Engineering}\\
\text{Feature Scaling \& Selection}
\end{cases}
\rightarrow
\text{Modeling}
$$

Then:

$$
\text{Modeling}
\rightarrow
\text{Model Evaluation \& Tuning}
\rightarrow
\text{Deployment \& Monitoring}
$$

If the model performance is not satisfactory:

$$
\text{Model Evaluation \& Tuning}
\rightarrow
\text{Data Preparation}
\rightarrow
\text{Modeling}
\rightarrow
\text{Evaluation \& Tuning}
$$

This process is **repeated until satisfactory model performance is achieved**.

# Feature Engineering

## What is Feature Engineering?

Feature Engineering is the process of **creating, transforming, selecting, or extracting features from raw data** so that the data becomes more useful for a Machine Learning algorithm.

The main areas of Feature Engineering are:

1. **Feature Transformation**
2. **Feature Construction**
3. **Feature Selection**
4. **Feature Extraction**

---

## Feature Engineering Overview

```text
                         Feature
                       Engineering
                            │
          ┌─────────────────┼─────────────────┬─────────────────┐
          │                 │                 │                 │
          ▼                 ▼                 ▼                 ▼
   Feature              Feature          Feature          Feature
 Transformation        Construction      Selection        Extraction
          │
          ├──→ Missing Value Imputation
          │
          ├──→ Handling Categorical Features
          │
          ├──→ Outlier Detection
          │
          └──→ Feature Scaling
```

---

# 1. Feature Transformation

Feature Transformation means **changing the representation or values of an existing feature** while keeping the underlying information.

The flowchart includes:

- Missing Value Imputation
- Handling Categorical Features
- Outlier Detection
- Feature Scaling

---

## 1.1 Missing Value Imputation

Missing Value Imputation means **replacing missing values with appropriate values** instead of leaving them empty.

### Example

Suppose we have:

| Age |
|-----|
| 20 |
| 25 |
| NaN |
| 30 |
| 35 |

We could replace the missing value using the median:

$$
\text{Median Age} = 25
$$

Therefore:

| Age |
|-----|
| 20 |
| 25 |
| 25 |
| 30 |
| 35 |

In Python:

```python
df["Age"] = df["Age"].fillna(df["Age"].median())
```

---

# 1.2 Handling Categorical Features

Machine Learning algorithms generally require numerical input.

Therefore, categorical variables often need to be **converted into numerical representations**.

### Example

Suppose:

| Sex |
|-----|
| Male |
| Female |
| Female |
| Male |

We can use encoding:

| Sex | Encoded |
|-----|--------:|
| Male | 0 |
| Female | 1 |

For categories without an inherent order, **One-Hot Encoding** can also be used.

Example:

| Sex | Male | Female |
|-----|-----:|-------:|
| Male | 1 | 0 |
| Female | 0 | 1 |

---

# 1.3 Outlier Detection

An outlier is an observation that is **unusually far from the rest of the data**.

### Example

Suppose the following values represent fares:

$$
10,\ 12,\ 15,\ 18,\ 20,\ 22,\ 500
$$

The value

$$
500
$$

is much larger than the other observations and may be an outlier.

Common techniques for detecting outliers include:

- Box Plot
- IQR
- Z-score
- Percentiles

The appropriate treatment depends on whether the value is a genuine observation or an error.

---

# 1.4 Feature Scaling

Feature Scaling changes the scale of numerical features so that features with very different ranges can be used more effectively by certain Machine Learning algorithms.

### Example

Suppose we have:

$$
\text{Age} = 18\text{--}80
$$

while:

$$
\text{Income} = 20,000\text{--}500,000
$$

Income has a much larger numerical scale.

One common technique is **Standardization**:

$$
z = \frac{x-\mu}{\sigma}
$$

where:

- $x$ = original value
- $\mu$ = mean
- $\sigma$ = standard deviation

# 2. Feature Construction

Feature Construction means **creating new features from existing features**.

The goal is to create information that may be more useful for the Machine Learning model.

### Example

Suppose a dataset contains:

| Height (cm) | Weight (kg) |
|-------------|------------:|
| 170 | 70 |
| 180 | 80 |

Instead of using only Height and Weight, we can construct a new feature:

$$
BMI = \frac{Weight}{Height^2}
$$

after converting height into meters.

For example:

$$
BMI = \frac{70}{1.70^2}
\approx 24.22
$$

Thus, we created a new feature:

```text
Height + Weight
        ↓
       BMI
```

Another Titanic example could be constructing:

```text
FamilySize = SibSp + Parch + 1
```

This creates a new feature representing the passenger's family size.

---

# 3. Feature Selection

Feature Selection means **selecting the most useful features and removing features that are unnecessary or harmful**.

We may remove features that are:

- Irrelevant
- Redundant
- Highly correlated with another feature
- Almost constant
- Potentially harmful to the model

### Example

Suppose our dataset contains:

```text
Age
Fare
Sex
PassengerId
```

`PassengerId` is simply an identifier and usually does not provide useful information about whether someone survived.

Therefore:

```text
Age
Fare
Sex
PassengerId
        ↓
Remove PassengerId
        ↓
Age
Fare
Sex
```

The goal is not to keep every available feature.

> **More features do not automatically mean a better Machine Learning model.**

---

# 4. Feature Extraction

Feature Extraction means **extracting useful information from existing raw data and representing it as features**.

This is especially useful when the original data contains complex information such as text, images, or dates.

### Example: Name

Suppose Titanic contains:

```text
"Cumings, Mrs. John Bradley"
```

Instead of using the entire name directly, we can extract the person's title:

```text
"Cumings, Mrs. John Bradley"
              ↓
             Mrs
```

Now `Mrs` becomes a categorical feature.

Another example is extracting information from a date:

```text
2026-08-10
```

can become:

```text
Year  → 2026
Month → 8
Day   → 10
```

Thus:

$$
\text{Raw Data}
\rightarrow
\text{Useful Features}
$$

---

# Feature Transformation vs Construction vs Selection vs Extraction

| Type | Main Idea | Example |
|------|-----------|---------|
| **Feature Transformation** | Change an existing feature | Standardize `Age` |
| **Feature Construction** | Create a new feature from existing features | `FamilySize = SibSp + Parch + 1` |
| **Feature Selection** | Keep useful features and remove unnecessary ones | Remove `PassengerId` |
| **Feature Extraction** | Extract useful information from raw/complex data | Extract `Mrs` from `Name` |

---

# Easy Mental Model

Think of Feature Engineering as asking four questions:

### 1. Can I improve an existing feature?

$$
\rightarrow \text{Feature Transformation}
$$

### 2. Can I create a useful new feature?

$$
\rightarrow \text{Feature Construction}
$$

### 3. Which features should I keep?

$$
\rightarrow \text{Feature Selection}
$$

### 4. Can I extract useful information from raw data?

$$
\rightarrow \text{Feature Extraction}
$$

Suppose the original dataset contains:

```text
Age
Sex
SibSp
Parch
Fare
Name
PassengerId
```

Feature Engineering might produce:

```text
Age
Sex_encoded
FamilySize
Fare
Title
```

while removing:

The process could look like:

```text
                     Raw Titanic Data
                            │
                            ▼
                   Feature Engineering
                            │
       ┌────────────────────┼────────────────────┐
       │                    │                    │
       ▼                    ▼                    ▼
 Transformation       Construction          Extraction
       │                    │                    │
       ▼                    ▼                    ▼
  Scale Age           FamilySize              Title
                      = SibSp + Parch + 1
       │
       ▼
  Handle Missing
       │
       ▼
  Encode Sex
       │
       └────────────────────┬───────────────────┘
                            ▼
                     Feature Selection
                            │
                            ▼
                    Remove PassengerId
                            │
                            ▼
                     Final Features
```

> **Key idea:** Feature Engineering takes **raw data** and tries to produce a representation that contains the information a Machine Learning model needs in a more useful form.
