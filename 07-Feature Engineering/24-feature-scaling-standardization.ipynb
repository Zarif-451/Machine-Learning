# Feature Scaling

Feature Scaling is the process of **transforming numerical features so that they are on a similar scale**.

This is important because different features may have very different ranges.

---

## Why is Feature Scaling Needed?

Suppose a dataset contains:

$$
\text{Age} = 18\text{--}80
$$

and

$$
\text{Salary} = 20,000\text{--}500,000
$$

The numerical values of `Salary` are much larger than those of `Age`.

For some Machine Learning algorithms, this difference in scale can cause the larger-valued feature to have an unnecessarily large influence.

Therefore, we can transform the features to comparable scales.

---

## Example

Before scaling:

| Age | Salary |
|----:|-------:|
| 20 | 30,000 |
| 30 | 80,000 |
| 40 | 150,000 |
| 50 | 300,000 |

After scaling, both features can be represented on a similar numerical scale.

```text
Age ────────┐
            ├──→ Feature Scaling ──→ Comparable Scales
Salary ─────┘
```

---

# Main Feature Scaling Techniques

The two techniques you should learn first are:

1. **Standardization**
2. **Normalization**

---

# 1. Standardization

Standardization transforms a feature so that its:

$$
\text{Mean} \approx 0
$$

and

$$
\text{Standard Deviation} \approx 1
$$

The formula is:

$$
z = \frac{x-\mu}{\sigma}
$$

where:

- $x$ = original value
- $\mu$ = mean of the feature
- $\sigma$ = standard deviation
- $z$ = standardized value

### Python

```python
from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()

X_scaled = scaler.fit_transform(X)
```

---

# 2. Normalization (Min-Max Scaling)

Normalization transforms values into a fixed range, commonly:

$$
[0,1]
$$

The Min-Max formula is:

$$
x' =
\frac{x-x_{\min}}
{x_{\max}-x_{\min}}
$$

where:

- $x$ = original value
- $x_{\min}$ = minimum value
- $x_{\max}$ = maximum value
- $x'$ = scaled value

### Python

```python
from sklearn.preprocessing import MinMaxScaler

scaler = MinMaxScaler()

X_scaled = scaler.fit_transform(X)
```

---

# Standardization vs Normalization

| Technique | Result | Formula |
|-----------|--------|---------|
| **Standardization** | Mean $\approx 0$, Std $\approx 1$ | $\frac{x-\mu}{\sigma}$ |
| **Normalization** | Usually values between $0$ and $1$ | $\frac{x-x_{\min}}{x_{\max}-x_{\min}}$ |

---

# When is Feature Scaling Important?

Scaling is particularly important for algorithms that depend on:

- Distance
- Magnitude
- Gradient-based optimization

Examples include:

- K-Nearest Neighbors (KNN)
- K-Means
- Support Vector Machines (SVM)
- Logistic Regression
- Linear Regression
- Neural Networks

---

# Example: Why Scaling Matters

Suppose we have two features:

$$
\text{Age} = 20
$$

and

$$
\text{Salary} = 100000
$$

A distance-based algorithm may calculate:

$$
\sqrt{(20)^2 + (100000)^2}
$$

The Salary value dominates the calculation.

After scaling, both features have comparable magnitudes, allowing the algorithm to consider them more fairly.

---

# Important Point

Feature scaling **does not change the underlying information** in a feature.

It changes its **numerical representation**.

For example:

```text
Original Age
     ↓
     25
     ↓
Standardization
     ↓
    -0.4
```

The person did not become younger.

Only the representation of the value changed.

---

# Easy Mental Model

Think:

```text
Different Feature Ranges
        ↓
Feature Scaling
        ↓
Similar Numerical Scale
        ↓
Some ML Algorithms Work Better
```

### Remember

$$
\boxed{
\text{Standardization}
=
\frac{x-\mu}{\sigma}
}
$$

$$
\boxed{
\text{Min-Max Normalization}
=
\frac{x-x_{\min}}
{x_{\max}-x_{\min}}
}
$$

> **Key idea:** Feature scaling is mainly about preventing differences in numerical scale from unfairly influencing algorithms that are sensitive to feature magnitude.

# Standardization

## What is Standardization?

**Standardization** is a Feature Scaling technique that transforms a numerical feature so that its:

$$
\text{Mean} \approx 0
$$

and

$$
\text{Standard Deviation} \approx 1
$$

It is also commonly called **Z-score normalization** or **Z-score standardization**.

The standardized value is called the **z-score**.

---

# Why Do We Need Standardization?

Different numerical features can have very different ranges.

For example:

| Feature | Range |
|---|---:|
| Age | 18--80 |
| Salary | 20,000--500,000 |
| Fare | 5--500 |

If an ML algorithm uses distances or numerical magnitudes, the feature with larger values can have a disproportionately large influence.

Standardization converts these features to a comparable scale.

---

# Standardization Formula

The standardization formula is:

$$
z = \frac{x-\mu}{\sigma}
$$

where:

- $x$ = original value
- $\mu$ = mean of the feature
- $\sigma$ = standard deviation of the feature
- $z$ = standardized value (z-score)

---

# Understanding the Formula

The formula:

$$
z = \frac{x-\mu}{\sigma}
$$

can be understood in two steps.

### Step 1: Subtract the Mean

$$
x-\mu
$$

This tells us how far the observation is from the mean.

### Step 2: Divide by Standard Deviation

$$
\frac{x-\mu}{\sigma}
$$

This expresses the distance from the mean in terms of **standard deviations**.

Therefore:

> A z-score tells us how many standard deviations an observation is above or below the mean.

---

# Example

Suppose the ages in a dataset have:

$$
\mu = 30
$$

and

$$
\sigma = 10
$$

Consider a person whose age is:

$$
x = 50
$$

Using the standardization formula:

$$
z = \frac{x-\mu}{\sigma}
$$

Substitute the values:

$$
z = \frac{50-30}{10}
$$

$$
z = \frac{20}{10}
$$

$$
\boxed{z=2}
$$

Therefore, an age of $50$ is **2 standard deviations above the mean**.

---

# Another Example

Suppose:

$$
\mu = 30
$$

$$
\sigma = 10
$$

and:

$$
x = 20
$$

Then:

$$
z = \frac{20-30}{10}
$$

$$
z = \frac{-10}{10}
$$

$$
\boxed{z=-1}
$$

Therefore, an age of $20$ is **1 standard deviation below the mean**.

---

# Interpreting Z-scores

| Z-score | Meaning |
|--------:|---------|
| $z=0$ | Exactly at the mean |
| $z=1$ | 1 standard deviation above the mean |
| $z=2$ | 2 standard deviations above the mean |
| $z=-1$ | 1 standard deviation below the mean |
| $z=-2$ | 2 standard deviations below the mean |

In general:

$$
z>0
$$

means the value is above the mean.

$$
z<0
$$

means the value is below the mean.

$$
z=0
$$

means the value is exactly equal to the mean.

---

# What Happens After Standardization?

Suppose we have:

$$
X = [10,20,30,40,50]
$$

The mean is:

$$
\mu = 30
$$

The population standard deviation is approximately:

$$
\sigma = 14.14
$$

For $x=10$:

$$
z=\frac{10-30}{14.14}
\approx -1.41
$$

For $x=20$:

$$
z=\frac{20-30}{14.14}
\approx -0.71
$$

For $x=30$:

$$
z=\frac{30-30}{14.14}
=0
$$

For $x=40$:

$$
z=\frac{40-30}{14.14}
\approx 0.71
$$

For $x=50$:

$$
z=\frac{50-30}{14.14}
\approx 1.41
$$

Therefore:

$$
X=[10,20,30,40,50]
$$

becomes approximately:

$$
Z=[-1.41,-0.71,0,0.71,1.41]
$$

Notice that the original values ranged from $10$ to $50$, while the standardized values are centered around $0$.

---

# What Does Standardization Actually Do?

Before:

$$
X: 10,\ 20,\ 30,\ 40,\ 50
$$

After:

$$
Z: -1.41,\ -0.71,\ 0,\ 0.71,\ 1.41
$$

The **relative ordering is preserved**:

$$
10 < 20 < 30 < 40 < 50
$$

becomes:

$$
-1.41 < -0.71 < 0 < 0.71 < 1.41
$$

The data has not been fundamentally changed.

Only its **scale and location** have changed.

---

# Important Property

After standardization:

$$
\text{Mean}(Z) \approx 0
$$

and:

$$
\text{Standard Deviation}(Z) \approx 1
$$

This is the main purpose of standardization.

---

# Standardization Does NOT Mean

Standardization does **not** mean:

> "Make every value between 0 and 1."

That is **Min-Max Normalization**.

### Standardization

$$
z=\frac{x-\mu}{\sigma}
$$

Typically produces values such as:

$$
-2.1,\ -0.5,\ 0,\ 1.3,\ 2.4,\ldots
$$

There is **no fixed minimum or maximum**.

### Min-Max Normalization

$$
x'=
\frac{x-x_{\min}}
{x_{\max}-x_{\min}}
$$

Usually produces values between:

$$
0 \leq x' \leq 1
$$

---

# Standardization in Python

Using Scikit-learn:

```python
from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()

X_scaled = scaler.fit_transform(X)
```

`StandardScaler` performs the transformation:

$$
z=\frac{x-\mu}{\sigma}
$$

for each numerical feature.

---

Suppose:

```python
df[["Age", "Fare"]]
```

| Age | Fare |
|----:|-----:|

```python
)
```

Now both features are represented on a comparable standardized scale.

---

# Why Is Standardization Useful in Machine Learning?

Standardization is particularly useful for algorithms that are sensitive to feature scale.

Examples include:

- K-Nearest Neighbors (KNN)
- K-Means Clustering
- Support Vector Machines (SVM)
- Logistic Regression
- Linear Regression
- Neural Networks
- Principal Component Analysis (PCA)

For example, consider:

$$
\text{Age}=25
$$

and:

$$
\text{Salary}=100000
$$

Without scaling, the numerical magnitude of Salary is much larger.

After standardization, both features are expressed in terms of their distance from their respective means.

---

# Standardization and Outliers

Standardization is **sensitive to outliers** because it uses:

$$
\mu
$$

and:

$$
\sigma
$$

Both the mean and standard deviation can be strongly affected by extreme values.

For example:

$$
10,\ 11,\ 12,\ 13,\ 1000
$$

The value $1000$ can significantly change the mean and standard deviation.

Therefore, if a feature contains severe outliers, we should investigate them before deciding how to scale the feature.

---

# Important: Standardization Does Not Remove Outliers

Suppose:

$$
z=5
$$

after standardization.

The observation is still an extreme observation.

Standardization only changes its representation.

It does **not automatically delete or fix outliers**.

---

# Important: Standardization Does Not Make Data Normal

This is a very common misconception.

Standardization gives approximately:

$$
\mu=0
$$

and:

$$
\sigma=1
$$

but it does **not** guarantee that the distribution becomes a normal distribution.

For example, if the original data is highly right-skewed, standardization will generally leave the distribution right-skewed.

---

# When Should I Use Standardization?

A useful rule is:

> If the Machine Learning algorithm is sensitive to feature scale, standardization is often useful.

It is especially important for algorithms based on:

$$
\text{Distance}
$$

or:

$$
\text{Magnitude}
$$

or:

$$
\text{Gradient Optimization}
$$

---

# When Is It Usually Not Necessary?

Tree-based algorithms generally do not require feature scaling.

Examples:

- Decision Tree
- Random Forest
- XGBoost
- LightGBM
- CatBoost

These algorithms split data based on feature values rather than calculating distances between observations.

---

```text
       ↓

       ↓
Standardized Feature
       ↓
Mean ≈ 0

```

Mathematically:

$$

}
$$

---

# Key Points to Remember

1. **Standardization is a Feature Scaling technique.**

2. It uses:

$$
z=\frac{x-\mu}{\sigma}
$$

3. The transformed feature has approximately:

$$
\mu=0
$$

and:

$$
\sigma=1
$$

4. A z-score tells us how many standard deviations a value is from the mean.

5. Positive $z$ → above the mean.

6. Negative $z$ → below the mean.

7. $z=0$ → exactly at the mean.

8. Standardization does **not** force values into $[0,1]$.

9. Standardization does **not** make a distribution normally distributed.

10. Standardization can be affected by outliers because it uses the mean and standard deviation.

11. It is particularly useful for scale-sensitive algorithms such as KNN, K-Means, SVM, and many gradient-based models.

> **The core idea:**
>
> $$\boxed{\text{Standardization expresses each value in units of standard deviation from the mean.}}$$

# Mean Centering

Mean centering is a technique in which the **mean of a feature is subtracted from every value** in that feature.

## Formula

$$
x_{\text{centered}} = x - \mu
$$

where:

- $x$ = original value
- $\mu$ = mean of the feature
- $x_{\text{centered}}$ = centered value

---

## Example

Suppose:

$$
X = [10,20,30,40,50]
$$

The mean is:

$$
\mu = \frac{10+20+30+40+50}{5}
$$

$$
\mu = 30
$$

Now subtract the mean from every value:

$$
10-30=-20
$$

$$
20-30=-10
$$

$$
30-30=0
$$

$$
40-30=10
$$

$$
50-30=20
$$

Therefore:

$$
X_{\text{centered}}
=
[-20,-10,0,10,20]
$$

The mean of the centered data is:

$$
\boxed{\text{Mean}(X_{\text{centered}})=0}
$$

---

# What Does Mean Centering Do?

Mean centering moves the **center of the data to zero**.

Before centering:

$$
X=[10,20,30,40,50]
$$

with:

$$
\mu=30
$$

After centering:

$$
X_{\text{centered}}
=
[-20,-10,0,10,20]
$$

with:

$$
\mu=0
$$

The relative distances between observations remain unchanged.

---

# Mean Centering vs Standardization

Mean centering only subtracts the mean:

$$
\boxed{x' = x-\mu}
$$

Therefore:

$$
\text{Mean}=0
$$

but the standard deviation is **not changed**.

---

Standardization goes one step further.

After subtracting the mean, we divide by the standard deviation:

$$
\boxed{
z=\frac{x-\mu}{\sigma}
}
$$

Therefore, standardization produces approximately:

$$
\text{Mean}=0
$$

and:

$$
\text{Standard Deviation}=1
$$

---

# Relationship

```text
Original Data
      ↓
Subtract Mean
      ↓
Mean Centering
      ↓
Mean = 0
      ↓
Divide by Standard Deviation
      ↓
Standardization
      ↓
Mean = 0
Std ≈ 1
```

Mathematically:

$$
\underbrace{x-\mu}_{\text{Mean Centering}}
$$

followed by:

$$
\frac{x-\mu}{\sigma}
$$

which gives:

$$
\underbrace{\frac{x-\mu}{\sigma}}_{\text{Standardization}}
$$

---

# Easy Way to Remember

> **Mean Centering:** Move the data so that its mean becomes $0$.

> **Standardization:** Move the mean to $0$ **and** scale the standard deviation to $1$.

### In short:

$$
\boxed{
\text{Mean Centering}
=
x-\mu
}
$$

$$
\boxed{
\text{Standardization}
=
\frac{x-\mu}{\sigma}
}
$$

# `StandardScaler`: `fit()` vs `transform()`

## The Process

```text
X_train
   ↓
 fit()
   ↓
Learn μ and σ
   ↓
transform X_train
   ↓
X_train_scaled

Same scaler
   ↓
transform X_test
   ↓
X_test_scaled
```

---

## `fit()`

The `fit()` method **learns the parameters from the training data**.

For `StandardScaler`, it calculates:

$$
\mu = \text{mean}
$$

and:

$$
\sigma = \text{standard deviation}
$$

Therefore:

$$
\boxed{
\text{fit}(X_{\text{train}})
\rightarrow
\text{learn } \mu \text{ and } \sigma
}
$$

Example:

```python
scaler.fit(X_train)
```

At this point, the scaler has learned the mean and standard deviation of each feature from `X_train`.

You can see the learned means using:

```python
scaler.mean_
```

---

## `transform()`

The `transform()` method uses the parameters learned during `fit()` to transform the data.

For Standardization:

$$
z = \frac{x-\mu}{\sigma}
$$

Therefore:

$$
\boxed{
\text{transform}
\rightarrow
\frac{x-\mu}{\sigma}
}
$$

Example:

```python
X_train_scaled = scaler.transform(X_train)
```

The same learned $\mu$ and $\sigma$ are used.

---

# Why Do We Fit Only on `X_train`?

The test set should represent **unseen data**.

Therefore, we should **not learn the scaling parameters from the test set**.

Correct:

```python
scaler.fit(X_train)

X_train_scaled = scaler.transform(X_train)
X_test_scaled = scaler.transform(X_test)
```

Incorrect:

```python
scaler.fit(X_test)
```

The test set should not influence the parameters learned during preprocessing.

---

# The Complete Process

```text
              X_train
                 │
                 ▼
              fit()
                 │
                 ▼
          Learn μ and σ
                 │
        ┌────────┴────────┐
        ▼                 ▼
transform X_train   transform X_test
        │                 │
        ▼                 ▼
X_train_scaled      X_test_scaled
```

The important point is that **the same scaler** is used for both datasets.

---

# `fit()` vs `transform()`

| Method | What it does |
|---|---|
| `fit()` | Learns parameters from the data |
| `transform()` | Uses learned parameters to transform data |
| `fit_transform()` | Performs `fit()` followed by `transform()` |

For `StandardScaler`:

$$
\boxed{
\text{fit} \rightarrow \text{learn } \mu,\sigma
}
$$

$$
\boxed{
\text{transform} \rightarrow \frac{x-\mu}{\sigma}
}
$$

---

# Common Shortcut

Instead of:

```python
scaler.fit(X_train)

X_train_scaled = scaler.transform(X_train)
```

we can use:

```python
X_train_scaled = scaler.fit_transform(X_train)
```

Then for the test set:

```python
X_test_scaled = scaler.transform(X_test)
```

So the standard pattern is:

```python
scaler = StandardScaler()

X_train_scaled = scaler.fit_transform(X_train)

X_test_scaled = scaler.transform(X_test)
```

### Remember

$$
\boxed{
\text{Training data}
\rightarrow
\text{fit\_transform}
}
$$

$$
\boxed{
\text{Test data}
\rightarrow
\text{transform only}
}
$$

> **Easy rule:**  
> **Fit on train, transform train and test using the same fitted scaler.**

# Train-Test Split

## What is Train-Test Split?

A **Train-Test Split** divides a dataset into two parts:

1. **Training Data** — used to train the Machine Learning model.
2. **Testing Data** — used to evaluate the model on data it has not seen before.

The basic idea is:

$$
\text{Dataset}
\rightarrow
\begin{cases}
\text{Training Data}\\
\text{Testing Data}
\end{cases}
$$

---

# Why Do We Split the Data?

Suppose we train a model using all of our available data and then evaluate it using the **same data**.

The model may perform extremely well because it has already seen those observations.

This does not tell us how well the model will perform on **new, unseen data**.

Therefore, we keep a portion of the data aside for testing.

Think of it like an examination:

```text
Learning Questions
       ↓
   Training Data
       ↓
      Model
       ↓
New Questions
       ↓
    Test Data
       ↓
   Evaluate Model
```

The testing data should represent data that the model has **not seen during training**.

---

# Separating Features and Target

Suppose our dataset contains:

| Age | EstimatedSalary | Purchased |
|----:|----------------:|----------:|
| 39 | 77000 | 0 |
| 37 | 80000 | 0 |
| 40 | 75000 | 0 |
| 47 | 30000 | 0 |
| 36 | 126000 | 1 |

Here:

```text
Age
EstimatedSalary
```

are the **features**.

And:

```text
Purchased
```

is the **target**.

# Why `random_state`?

The train-test split is random.

If we run:

```python
train_test_split(X, y, test_size=0.3)
```

the rows selected for training and testing can change each time.

We can make the split **reproducible** by specifying:

```python
random_state=0
```

For example:

```python
train_test_split(
    X,
    y,
    test_size=0.3,
    random_state=0
)
```

The value `0` is not special.

We could also use:

```python
random_state=42
```

or:

```python
random_state=10
```

The important idea is:

> Using the same `random_state` gives the same random split when the other inputs are unchanged.

# When to Use Feature Scaling

Feature scaling is particularly important for the following algorithms:

| Algorithm | Reason for Applying Feature Scaling |
|---|---|
| **K-Means** | Uses the **Euclidean distance** measure. |
| **K-Nearest Neighbours (KNN)** | Measures distances between pairs of samples, and these distances are influenced by the measurement units. |
| **Principal Component Analysis (PCA)** | Attempts to find features/components with **maximum variance**. |
| **Artificial Neural Networks (ANN)** | Uses **Gradient Descent**, which benefits from appropriately scaled features. |
| **Gradient Descent** | Feature scaling can make parameter updates faster and improve the optimization process. |

---

## Easy Way to Remember

Feature scaling is especially important when an algorithm depends on:

$$
\boxed{\text{Distance}}
$$

$$
\boxed{\text{Variance}}
$$

or

$$
\boxed{\text{Gradient Descent}}
$$

Therefore, remember:

```text
K-Means       → Distance
KNN           → Distance
PCA           → Variance
Neural Network → Gradient Descent
Gradient Descent → Optimization
```

> **Key idea:** Algorithms that are sensitive to the magnitude or scale of features generally benefit from feature scaling.
